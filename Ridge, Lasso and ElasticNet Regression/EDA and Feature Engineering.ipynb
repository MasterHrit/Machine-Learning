```python
df["Classes"]=np.where(df["Classes"].str.contains("not fire"),"not fire","fire")
```

```python
df.head()
```

```python
## Monthly Fire Analysis
dftemp=df[df["Region"]==1]
plt.subplots(figsize=(10,6))
sns.set_style("whitegrid")
sns.countplot(x="month",hue="Classes",data=dftemp)
plt.xlabel("Month",weight="bold")
plt.ylabel("Number of Fires", weight="bold")
plt.title("Fire Analysis for Siddi Bell Region",weight="bold")
plt.show()
```

```python
## Monthly Fire Analysis
dftemp=df[df["Region"]==0]
plt.subplots(figsize=(10,6))
sns.set_style("whitegrid")
sns.countplot(x="month",hue="Classes",data=dftemp)
plt.xlabel("Month",weight="bold")
plt.ylabel("Number of Fires", weight="bold")
plt.title("Fire Analysis for Bejaia Region",weight="bold")
plt.show()
```

```python
# ## Most of the fire happened in August and then in Jul, in both the regions.
# Least Fire happened in September for both region.
# We can also see that in Bejaia Region most of the rainfall also occured


## pair plot
sns.pairplot(df_copy)
```

```python
## Distribution for Temperature for each region monthly. mean ?
df.head()
```

```python
## For Region Bejaia
dftemp=df[df["Region"]==0]
mean_temp_grouped_monthly=dftemp.groupby("month")["Temperature"].mean()
mean_temp_grouped_monthly
```

```python
plt.bar(mean_temp_grouped_monthly.index,mean_temp_grouped_monthly)
plt.grid(visible=False,axis="x")
plt.xticks(mean_temp_grouped_monthly.index)
plt.xlabel("Month",weight="bold")
plt.ylabel("Mean Temperature",weight="bold")
plt.title("Mean Temperature per Month for Region Bejaia",weight="bold")
plt.show()
```

```python
## For Region Sidi-Bel Abbes
dftemp=df[df["Region"]==1]
mean_temp_grouped_monthly=dftemp.groupby("month")["Temperature"].mean()

plt.bar(mean_temp_grouped_monthly.index,mean_temp_grouped_monthly)
plt.grid(visible=False,axis="x")
plt.xticks(mean_temp_grouped_monthly.index)
plt.xlabel("Month",weight="bold")
plt.ylabel("Mean Temperature",weight="bold")
plt.title("Mean Temperature per Month for Region Sidi-Bel Abbes",weight="bold")
plt.show()
```

```python
# From the above graphs for both region we can see the temperature was highest in 8th Month i.e August and Secondly in July. Whereas least being in the September
```
