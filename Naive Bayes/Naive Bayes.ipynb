***Naive Bayes Algorithm Implementation***

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
load_iris()
```

```python
X,y=load_iris(return_X_y=True)
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
X_train
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
y_pred=gnb.predict(X_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
# Predict time -- Just see how it will perform (after feature engineering)
```

```python
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(drop="first")
df["sex"]=ohe.fit_transform(df[["sex"]]).toarray()
```

```python
df.head()
```

```python
ohe.get_feature_names_out()
```

```python
df["smoker"]=ohe.fit_transform(df[["smoker"]]).toarray()
```

```python
df.head()
```

```python
df["day"].value_counts()
```

```python
encoded_day=ohe.fit_transform(df[["day"]]).toarray()
```

```python
import pandas as pd
day_df=pd.DataFrame(encoded_day,columns=ohe.get_feature_names_out())
```

```python
day_df.head()
```

```python
df=pd.concat([df,day_df],axis=1)
```

```python
df.drop(columns=["day"],inplace=True)
```

```python
df.head()
```

```python
df["time"]=ohe.fit_transform(df[["time"]]).toarray()
df.head()
```

```python
df["time"].value_counts()
```

```python
list(df.columns[:4])+list(df.columns[5:])
```

```python
## Independent and Dependent Separation
X=df[list(df.columns[:4])+list(df.columns[5:])]
y=df["time"]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=10)
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
```

```python
gnb.fit(X_train,y_train)
```

```python
y_pred=gnb.predict(X_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
logistic.fit(X_train,y_train)
y_pred=logistic.predict(X_test)
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
from sklearn.svm import SVC
svc=SVC(kernel="linear") # Other kernels were not giving good accuracy
svc.fit(X_train,y_train)
y_pred=svc.predict(X_test)
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```
