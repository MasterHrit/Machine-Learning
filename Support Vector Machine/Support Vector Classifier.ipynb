**SVC Implementation**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")
```

```python
## Create synthetic datapoints -- for solving classification problem
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=1,n_redundant=0)
```

```python
X
```

```python
y
```

```python
pd.DataFrame(X)[0]
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0],y=pd.DataFrame(X)[1],hue=y) 
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
from sklearn.svm import SVC
svc=SVC(kernel="linear")
```

```python
svc.fit(X_train,y_train)
```

```python
## Prediction
y_pred=svc.predict(X_test)
```

```python
plt.subplots(figsize=(13,6))
plt.subplot(1,2,1)
sns.scatterplot(x=pd.DataFrame(X_train)[0],y=pd.DataFrame(X_train)[1],hue=y_train)
plt.title("Actual")
plt.subplot(1,2,2)
sns.scatterplot(x=pd.DataFrame(X_train)[0],y=pd.DataFrame(X_train)[1],hue=svc.predict(X_train))
plt.title("Predicted")
plt.tight_layout()
plt.show()
```

```python
svc.coef_
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
# Defining parameter range
param_grid={"C":[0.1,1,10,100,1000],
            "gamma":[1,0.1,0.01,0.001,0.0001],
            "kernel":["rbf"]}
```

```python
grid=GridSearchCV(SVC(),param_grid=param_grid,cv=5,verbose=3,refit=True)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred4=grid.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(y_test,y_pred4))
print(confusion_matrix(y_test,y_pred4))
```
