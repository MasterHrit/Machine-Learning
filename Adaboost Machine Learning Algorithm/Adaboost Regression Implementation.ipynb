**Used Car Prediction**
1. Problem Statement
    - This dataset contains used cars sold on cardekho.com in India as well as important features of these cars
    - If user can predict the price of the car based on the input features
    - Prediction results can be used to give new seller the price suggestion based on market condition

2. Data Collection
    - The dataset is collected from scrapping from cardekho website
    - The data consists of 13 column and 15411 rows.

```python
## We will not transform the discrete numerical features 'vehicle_age', 'seats'
## Standard Scale 'km_driven', 'mileage', 'engine', 'max_power'
## OneHotEncode 'seller_type', 'fuel_type', 'transmission_type'
## LabelEncode 'model'
```

```python
# Lets first define the dependent and independent features and split the dataset
X=df.iloc[:,:-1]
y=df["selling_price"]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
X_train.shape,X_test.shape
```

```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
le=LabelEncoder()
X_train["model"]=le.fit_transform(X_train["model"])
X_test["model"]=le.transform(X_test["model"])
ss_features=['km_driven', 'mileage', 'engine', 'max_power']
ohe_features=['seller_type', 'fuel_type', 'transmission_type']
transformer=[
    ("StandardScaler",StandardScaler(),ss_features),
    ("OneHotEncoder",OneHotEncoder(drop="first"),ohe_features)
]

ct=ColumnTransformer(transformers=transformer,remainder='passthrough',n_jobs=-1,verbose=True)
```

```python
X_train=ct.fit_transform(X_train)
X_test=ct.transform(X_test)
```

```python
ct.get_feature_names_out()
```

```python
pd.DataFrame(X_train)
```

```python
## Model Training
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error
```

```python
## Create a Function to Evaluate Model
def evaluate_model(true,predicted):
    mae=mean_absolute_error(true,predicted)
    mse=mean_squared_error(true,predicted)
    rmse=root_mean_squared_error(true,predicted)
    r2=r2_score(true,predicted)
    return mae,rmse,r2
```

```python
## Model Training
models={
    "Linear Regression":LinearRegression(),
    "Lasso":Lasso(),
    "Ridge":Ridge(),
    "KNeighborsRegressor":KNeighborsRegressor(),
    "DecisionTreeRegressor":DecisionTreeRegressor(),
    "RandomForestRegressor":RandomForestRegressor(),
    "AdaBoostRegressor":AdaBoostRegressor()
}

for k,v in models.items():
    # Train Model
    v.fit(X_train,y_train)

    # Make predictions
    y_train_pred=v.predict(X_train)
    y_test_pred=v.predict(X_test)

    # Performance Check
    print(k)
    print("-"*30)
    print("Performance of Training Dataset : (mae,rmse,r2)")
    print(evaluate_model(y_train,y_train_pred))
    print("Performance of Test Dataset : (mae,rmse,r2)")
    print(evaluate_model(y_test,y_test_pred))
    print("\n")
```

```python
## Initialize parameter for Hyperparameter Tuning for both KNearestNeighbour and RandomForestRegressor
knn_param={
    "n_neighbors":[2,3,10,20,40,50]
}
rf_params={
    "n_estimators":[100,200,500,1000],
    "criterion":["squared_error", "absolute_error", "poisson"],
    "max_depth":[5,8,10,15,"None"],
    "min_samples_split":[2,8,15,20],
    "max_features":[5,7,"auto",8]
}
ada_params={
    "n_estimators":[50,60,70,80,90,100],
    "loss":["linear", "square", "exponential"]
}
```

```python
# Model List for Hyperparameter Tuning
randomcv_models=[
    ("KNeighboursRegressor",KNeighborsRegressor(),knn_param),
    ("RandomForestRegressor",RandomForestRegressor(),rf_params),
    ("AdaBoostRegressor",AdaBoostRegressor(),ada_params)
]
```

```python
from sklearn.model_selection import RandomizedSearchCV
model_param={}
for name,model,param in randomcv_models:
    random=RandomizedSearchCV(estimator=model,param_distributions=param,n_iter=100,cv=3,verbose=2,n_jobs=-1)
    random.fit(X_train,y_train)
    model_param[name]=random.best_params_

for model,param in model_param.items():
    print(f"Best Parameter for {model} :")
    print(param)
```

```python
## Model Training
models={
    "KNeighborsRegressor":KNeighborsRegressor(n_neighbors=2,n_jobs=-1),
    "RandomForestRegressor":RandomForestRegressor(n_estimators=500, min_samples_split=2, max_features=5, max_depth=15, criterion="absolute_error",n_jobs=-1),
    "AdaboostRegressor":AdaBoostRegressor(n_estimators=50,loss="linear")
}

for k,v in models.items():
    # Train Model
    v.fit(X_train,y_train)

    # Make predictions
    y_train_pred=v.predict(X_train)
    y_test_pred=v.predict(X_test)

    # Performance Check
    print(k)
    print("-"*30)
    print("Performance of Training Dataset : (mae,rmse,r2)")
    print(evaluate_model(y_train,y_train_pred))
    print("Performance of Test Dataset : (mae,rmse,r2)")
    print(evaluate_model(y_test,y_test_pred))
    print("\n")
```
